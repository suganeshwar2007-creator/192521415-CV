```

# Basic Image Handling and Processing
# Read an image and convert it to Grayscale

import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Step 1: Upload an image
uploaded = files.upload()

# Step 2: Read the uploaded image
filename = next(iter(uploaded))
image = cv2.imread(filename)

# Check whether the image was read successfully
if image is None:
    print("Error: Unable to read the image.")
else:
    # Step 3: Convert BGR image to RGB for display
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Step 4: Convert the image to grayscale
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Step 5: Display the original and grayscale images
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(image_rgb)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(gray_image, cmap="gray")
    plt.title("Grayscale Image")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    # Step 6: Save the grayscale image
    cv2.imwrite("grayscale_image.jpg", gray_image)
    print("Grayscale image saved successfully!")

    # Step 7: Download the processed image
    files.download("grayscale_image.jpg")
```
